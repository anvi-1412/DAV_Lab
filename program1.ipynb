```python
import numpy as np
import pandas as pd

arr = np.array([20, 57, 59, 97, 56])
print("array:", arr)
print("mean:", arr.mean())
print("reshaped 2D:\n", arr.reshape(5, 1))
print("sliced(index 1 to 3):", arr[1:4])


data = {
    "Name": ['anjali', 'akshatha', 'anagha', 'ashiq'],
    "Marks":[90,89,78,68],
    "Branch": ['CSE', 'CSE', 'CSE', 'CSE'],
}

df = pd.DataFrame(data)
print("\n First two rows:\n", df.iloc[0:2])
print("\n Marks column\n", df['Marks'])
print("\n Rows where Marks>70:\n", df[df['Marks'] > 70])
```
