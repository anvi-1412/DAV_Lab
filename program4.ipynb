```python
# Import Pandas for creating and manipulating DataFrames
import pandas as pd

# Import NumPy to represent missing numerical values using np.nan
import numpy as np


# --------------------------------------------------
# STEP 1: CREATE A DATASET WITH MISSING VALUES
# --------------------------------------------------

# Create a DataFrame containing property details.
# np.nan represents a missing value.
df = pd.DataFrame({
    "Area": [1200, 1500, np.nan, 1800, 2000],
    "Price": [45, 60, 55, np.nan, 95],
    "Location": ["Mysuru", "Hassan", np.nan,
                 "Mysuru", "Bengaluru"]
})

# Display the original dataset before handling missing values
print("Original Dataset:")
print(df)


# --------------------------------------------------
# STEP 2: IDENTIFY MISSING VALUES
# --------------------------------------------------

# isnull() identifies missing values in each cell.
# sum() counts the missing values column by column.
print("\nMissing Values in Each Column:")
print(df.isnull().sum())


# --------------------------------------------------
# STEP 3: DROP ROWS CONTAINING MISSING VALUES
# --------------------------------------------------

# dropna() removes rows containing one or more missing values.
# The result is stored in a new DataFrame called df_dropped.
# The original DataFrame df remains unchanged.
df_dropped = df.dropna()

# Display the dataset after removing incomplete rows
print("\nDataset After Dropping Rows:")
print(df_dropped)


# --------------------------------------------------
# STEP 4: MEAN IMPUTATION
# --------------------------------------------------

# Calculate the mean of the available Area values.
# fillna() replaces the missing Area value with that mean.
df["Area"] = df["Area"].fillna(
    df["Area"].mean()
)

# Display the Area column after mean imputation
print("\nArea After Mean Imputation:")
print(df["Area"])


# --------------------------------------------------
# STEP 5: MEDIAN IMPUTATION
# --------------------------------------------------

# Calculate the median of the available Price values.
# fillna() replaces the missing Price value with the median.
# The median is often useful when numerical data is skewed
# or contains outliers.
df["Price"] = df["Price"].fillna(
    df["Price"].median()
)

# Display the Price column after median imputation
print("\nPrice After Median Imputation:")
print(df["Price"])


# --------------------------------------------------
# STEP 6: MODE IMPUTATION
# --------------------------------------------------

# mode() finds the most frequently occurring Location.
# [0] selects the first mode if multiple modes exist.
# fillna() replaces the missing Location with that mode.
df["Location"] = df["Location"].fillna(
    df["Location"].mode()[0]
)

# Display the Location column after mode imputation
print("\nLocation After Mode Imputation:")
print(df["Location"])


# --------------------------------------------------
# STEP 7: FORWARD FILL
# --------------------------------------------------

# Create a time-series-like sequence of readings.
# The third reading is missing.
readings = pd.Series([28, 29, np.nan, 30])

# Display the readings before forward fill
print("\nReadings Before Forward Fill:")
print(readings)

# ffill() fills a missing value with the previous
# available value in the sequence.
# Here, the missing value becomes 29.
readings = readings.ffill()

# Display the readings after forward fill
print("\nReadings After Forward Fill:")
print(readings)


# --------------------------------------------------
# STEP 8: VERIFY THE FINAL DATASET
# --------------------------------------------------

# Display the dataset after mean, median and mode imputation
print("\nFinal Dataset After Imputation:")
print(df)

# Count missing values in every column again.
# Each count should be zero if all missing values were handled.
print("\nMissing Values After Imputation:")
print(df.isnull().sum())
```
